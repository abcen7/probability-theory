## Определение условной вероятности.
> Условной вероятностью $A$ при условии $B$, причем $P(B) > 0$, называется $P(A|B) = \frac{P(AB)}{P(B)}$ равносильно $(P_B(A))$. Здесь $P(AB)$ – вероятность одновременного наступления событий $A$ и $B$.

## Формула полной вероятности. 
> Пусть опыт проводится в два этапа.
> - Первый этап – происходит ровно одно из $n$ попарно несовместных событий ${H_1, H_2, H_3, \dots, H_n}$ образующих полную группу $\bigcup\limits_{i=1}^{n}H_i = \Omega$. Вероятность каждого из них положительная $(P(H_i))$. Они называются гипотезами.
> - Второй этап – после того, как произошло событие $H_i$, где $i \in {1, 2, ..., n}$ (принята гипотеза $H_i$), происходит или не происходит событие $A$

> Формула полной вероятности. Если событие $A$ может произойти только при условии появления одного из событий (гипотез) {${H_1, H_2, \dots, H_n}$}, образующих полную группу, то вероятность события $A$ равна сумме произведений вероятностей каждого из этих событий (гипотез) на соответствующие условные вероятности события $A$.
$$P(A) = \sum_{i=1}^nP(H_i) \cdot P(A|H_i)$$

### Примеры на ФПВ

### Задача про револьверы (Джон)

Рассмотрим событие $X$ - револьвер пристрелянный, $P(X) = 0.3$. $\bar{X}$ - не пристрелянный. $P(\bar{X}) = 0.7$. $Y$ – муха убита. $P_X(Y) = 0.8, P_{\bar{X}}(Y) = 0.4$. По ФПВ (Формула полной вероятности): $P(Y) = P(X) \cdot P_X(Y) + P(\bar{X}) \cdot P_{\bar{X}}(Y) = 0.3 \cdot 0.8 + 0.7 \cdot 0.4 = 0.52$

### Задача про яйца

Рассмотрим событие $O$ - яйцо произведено "Окское". $P(O) = 0.5$. $D$ - яйцо производителя "Деревенское". $P(D) = 0.3$. $K$ - яйцо производителя "Красная цена". $P(K) = 0.2$. $N$ – яйцо нулевой категории. По условию, $P_O(N) = 0.6$, $P_D(N) = 0.7$, $P_K(N) = 0.5$. По ФПВ: $P(N) = P(O) \cdot P_O(N) + P(D) \cdot P_D(N) + P(K) \cdot P_K(N) = 0.5 \cdot 0.6 + 0.3 \cdot 0.7 + 0.2 \cdot 0.5$

## Формула Байеса. 

Формула Байеса в простейшем варианте для двух случайных событий $A$ и $B$. 

$$P_B(A) = \frac{P(A) \cdot P_A(B)}{P(B)}.$$

$$P(A | B) = \frac{P(B | A) \cdot P(A)}{P(B)}.$$

> Априорная (т.е. имевшаяся **до** опыта) вероятность события $H_i$ – это вероятность $P(H_i)$

> Апостериорная (т.е. полученная **после** опыта) вероятность события $H_i$ – это условная вероятность $P(H_i|A)$ 

Формула Байеса позволяет вычислить апостериорную вероятность каждой из гипотез через априорные вероятности всех гипотез и условные вероятности рассматриваемого события, при условии принятия каждой из гипотез
