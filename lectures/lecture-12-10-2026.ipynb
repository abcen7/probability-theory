# 12-10-2026 Дискретные случайные величины и их характеристики

## Понятие ДСВ

> **Случайной величиной** называются величину, которая в результате испытания примет только одно возможное значение, наперед не известное и зависящее от случайных причин, которые заранее не известны

Пример №1. Число родившихся мальчиков из $100$ новорожденных – случайная величина, которая принимает значения: $0, 1, 2, \dots, 100$

Пример №2. Расстояние, которое пролетит снаряд при выстреле из орудия – случайная величина. Ее возможные значения принадлежат множеству $(a; b)$

> Дискретной называют СВ, которая принимает отдельные изолированные значения, с определенными вероятностями. Число возможных значений ДСВ может быть конечно или бесконечно. Непрерывной случайной величиной называют СВ, которая может принимать все значения из некоторого конечного или бесконечного промежутка. Число возможных значений бесконечно. 

Случайная величина $X$ есть функция $X = f(w)$ заданная на множестве элементарных исходов

## Закон распределения СВ

> Законом распределения СВ называется всякое соотношение, устанавливающее связь между возможными значениями случайной величины и соответствующими ей вероятности.

Для ДСВ закон распределения может быть задан в виде:

- таблицы - ряда распределения ДСВ
- графически - в виде многоугольника распределения

> В таблице по возрастанию указывается всевозможные значения случайной величины и соответствующие им вероятности. Такая таблица называется **рядом распределения ДСВ**

Событие $X = x_i$ состоит в том, что в результате испытания случайная величина $X$ примет значение $x_i$

Для любой ДСВ:

$$\sum\limits^{n}_{i=1}P(X = x_i) = \sum\limits^{n}_{i = 1}p_i = 1$$

### Пример. 

Построить ряд распределения и полигон распределения выигрыша: В лотерее выпущено 100 билетов. Разыгрывается один выигрыш в 50 у. е. и 10 выигрышей по 1 у. е.

Рассмотрим случайную величину $X$ – стоимость выигрыша на 1 билет

$$
\begin{array}{|c|c|c|c|}
\hline
x_i & 0 & 1 & 50 \\
\hline
p_i & 0.89 & 0.1 & 0.01 \\
\hline
\end{array}
$$

Проверка $\sum{p_i} = 1$

### Функция распределения вероятности СВ.

Пусть $X \in \mathbb{R}$. Вероятность того, что случайная величина $X$ примет значение $X < x$ обозначим $F(x)$, т.к. $x$ изменяется, то и $F(x)$ изменяется, т.е. $F(x)$ – функция от $x$
