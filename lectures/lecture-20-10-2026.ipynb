$F(x) $

Функцию распределения еще называют интегральной функцией

Свойства:

1. Функция распредления СВ - неотрицательная функция, заключенная между $0$ и $1$
2. Функция распределения СВ неубывающая на всей числовой оси
3. Если возможное значение НСВ расположены на всей числовой оси то справедливы: $$\lim_{x \to -\infty} F(x) = 0 \lim_{x \to -\infty} F(x) = 1$$
4. Вероятность попадания СВ в интервал $[x_1, x_2)$ равна приращению функции распределения на этом интервале $$P(x_1 \leq X < x_2) = F(x_2) - F(x_1)$$
5. Если возможные значения СВ принадлежат интервалу $(a, b)$, то $$F(x) = 0, \text{при } x \leq a$$ $$F(x) = 1, \text{при } x \geq b$$

График функции распределения ДСВ имеет ступенчатый вид.

### Пример. ДСВ $X$ задана таблицей распределения

$$
\begin{array}{|c|c|c|c|}
\hline
x_i & 1 & 5 & 8 \\
\hline
p_i & 0.3 & 0.1 & 0.6 \\
\hline
\end{array}
$$

$$
F(x) = 
\begin{cases}
0, & x \leq 1 \\
0 + 0.3 = 0.3, & 1 < x \leq 4 \\
0 + 0.3 + 0.1 = 0.4, & 4 < x \leq 8 \\
0 + 0.3 + 0.1 + 0.6 = 1, & x > 8 \\
\end{cases}
$$

$$

> Математическим ожиданием (средним значением) $M(X)$ дискретной случайной величины $X$ называется сумма произведений всех ее значений на соответствующие им вероятности

$$M(X) = \sum\limits^{n}_{i = 1}x_i \cdot p_i$$

### Пример. Найти $M(X)$

$$
\begin{array}{|c|c|c|c|}
\hline
x_i & 2 & 3 & 5 \\
\hline
p_i & 0.3 & 0.1 & 0.6 \\
\hline
\end{array}
$$

```python
xi = [2, 3, 5]
pi = [0.3, 0.1, 0.6]
p = 0
for i in range(3):
    p += xi[i] * pi[i]
p
```

Вероятностный смысл $\approx$ среднему арифметическому наблюдаемых значений. 

### Свойства:

1. $M(C) = C$
2. $M(kX) = kM(X)$
3. $M(X \pm Y) = M(X) \pm M(Y)$
4. Мат. ожидание проивездения конечного числа независимых случайных величин равно произведению их мат. ожиданий $M(X \cdot Y) = M(X) \cdot M(Y)$
5. $M(X \pm C) = M(X) \pm C$
6. $M(X - M(X)) = 0$

## Дисперсия

Дисперсия - мера разрбоса. На сколько сильно СВ отличается от мат. ожидания? 

Чтобы оценить, как рассеяны возможные значения СВ вокруг ее мат. ожидания используют дисперсию

> Дисперсией $D(X), V(X)$ случайной величины $X$ называется мат. ожидание квадрата ее отклонения от мат. ожидания.

$$D(X) = M([X - M(x)]^2)$$

### Пример. 

Найти дисперсию СВ $X$, которая задана законом распределения 

$$
\begin{array}{|c|c|c|c|}
\hline
x_i & 1 & 3 & 5 \\
\hline
p_i & 0.3 & 0.1 & 0.2 \\
\hline
\end{array}
$$

```python
x = [1, 2, 5]
ps = [0.3, 0.5, 0.2]
e = 0
for i in range(3):
    e += x[i] * ps[i]
e
```

```python
x_correlated = []
for i in range(3):
    x_correlated.append(x[i] - e)
x_correlated_squared = list(map(lambda x: x ** 2, x_correlated))
x_correlated, x_correlated_squared, 
```

```python
d = 0
for i in range(3):
    d += ps[i] * x_correlated_squared[i]
d
```

Для более удобного расчета используется следующая теорема:
$$D(X) = M(X^2) - (M(X))^2$$

```python
squared_x = list(map(lambda x: x ** 2, [1, 2, 5]))
squared_x
```

```python
e_squared_x = 0
for i in range(3):
    e_squared_x += squared_x[i] * ps[i]
e_squared_x - e ** 2
```

### Свойства дисперсии

1. $D(C) = 0$
2. $D(kX) = k^2D(X)$
3. $D(X + Y) = D(X) + D(Y)$
4. $D(X - Y) = D(X) + D(Y)$ - из-за квадрата в ее определении

```python
from scipy.stats import binom

p = 0.9
q = 1 - p
n = 5

for m in range(0, n + 1):
    print(m, n, f'{binom.pmf(m, n, p):.5f}')
```

Мат. ожидание $M(X)$ и дисперсия $D(X)$ числа появляения событий $A$ в $n$ независимых испытаниях равно

$$M(X) = np$$

$$D(X) = npq$$

```python
p = 0.6
q = 1 - p
n = 10
probabilities = []
for m in range(n + 1):
    calculated = binom.pmf(m, n, p)
    print(m, n, p, f'{calculated:.5f}')
    probabilities.append(calculated)
```

Дисперсия **не является** удобной мерой рассеивания СВ, поскольку измеряется в квадратных единицах, поэтому логично извлечь из нее корень 

> Средним квадратическим отклонением $\sigma(X)$ случайной величиной $X$ называют квадратный корень из дисперсии:

$$\sigma(X) = \sqrt{D(X)}$$
