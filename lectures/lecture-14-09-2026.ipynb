## Лекция №2

Частотный подход к определению вероятности. 

### Основные определения. 

> Случайный эксперимент (испытание) – любая процедура с несколькими конечными результатами, повторимая конечное или бесконечное число раз. Пример: "подбрасывание игральной кости"

> Любой результат эксперимента ($\omega \in \Omega$) называется элементарный исход

> Множество всевозможных исходов называется пространством элементарных исходов, обозначется $\Omega$. Пример: при подбрасывании игральной кости $\Omega = {1, 2, 3, 4, 5, 6}$

> Случайное событие – любое подмножество исходов эксперимента, обозначаются $A, B, C, D, ...$

Виды событий: несовместные (вместе произойти не могут), попарно-несовместные (никакие два из них не могут произойти одновременно), совместные (могут вместе прозойти), несовместные в совокупности

> Полная группа событий – совокупность событий, хотя бы одно из них произойдет

Подбрасывание двух монет: $\Omega = {OO, PO, OP, PP}$

### Подходы к определению вероятности. 

> Классический - рассматривается конечное число исходов, считается, что все исходы равновероятны и в этом подходе вероятность события $A$ определяется как отношение $N_A$ – количества благоприятных исходов, к общему количеству исходов $N$. $P(A) = \frac{N_A}{N}$

> Частотный – вычисление вероятности построено на эксперименте, который проводится $N$ раз ($N \rightarrow \infty $), а вероятность события $A$ определяется как отношение $N_A$ – количества исходов, когда событие $A$ произошло

> Байесовский – $\exists$ **априорная** вероятность, которая выражает степень нашей веры, то что этот исход произойдет, которая обновляется после эксперимента по формуле Байеса. Эта вероятность в бесконечном пределе к реальной вероятности

### Аксиоматика Колмогорова

Аксиома Колмогорова пытаются сделать переход от событий к вероятности. Цель – описать вероятность, как функцию мер. Правильно – $P(A) = 0.4$. Неправильно – $P = 0.4$. Только на вход подается пространство элементарных событий, а измеряем элементарные события. Колмогоров при переводе вероятности на математический язык, воспользовался теорией мер. Взял теорию мер и добавил очевидные свойства вероятности

Теория мер. 

**Сигма**-алгеброй $\mathcal{F}$ называется пространство $\Omega$, если он удовляетворяет следующим аксиомам:

1. $ \emptyset \in \mathcal{F}$
2. $\mathcal{F}$ замкнуто относительно дополнения
3. $\mathcal{F}$ замкнуто относительно счетного числа объединений: $A_1, A_2, ... \in \mathcal{F} \Rightarrow $

Если сигма-алгебра удовлетворяет аксиомам 1-3, то пространство элементарных событий, снабженных этой сигма-алгеброй, называется **измеримым пространством**.

Если функция $P$ удовлетворяет этимым аксиомам, то тройка $(\Omega, \mathcal{F}, P)$ называется вероятностным пространством, а $P(A)$ называется вероятностью события $A \in \mathcal{F}$

Аксиомы приводят к тому, что $P(A) \in [0, 1]$

### Сложение и умножение вероятностей

Два события называются **независимыми**, если вероятность наступления одного из них не влияет на вероятность наступления другого

Теорема **умножения** вероятностей: 

$P(A \cdot B) = \begin{cases}\
P(A) \cdot P(B), & A, B \text{ – независимые},\\
P(A) \cdot P_A(B) & A, B \text{ – зависимые}
\end{cases}
$
где $P_A(B)$ или $P(B | A)$ – вероятность события $B$, что $A$ – произошло

Правило **сложения** двух событий:

$P(A + B) = \begin{cases}\
P(A) + P(B), & A, B \text{ – несовместные},\\
P(A) + P(B) - P(A \cdot B) & A, B \text{ – совместные}
\end{cases}
$
