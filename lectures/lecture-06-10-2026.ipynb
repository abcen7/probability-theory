### Следствия из интегральной теоремы Муавра-Лапласа (позволяют найти отклонение частоты от вероятности)

а) если вероятность $p$ наступления события $A$ в каждом испытании постоянна и отлична от $0$ и $1$, то при достаточно большом числе $n$ независимых испытаний: вероятность того, что число $m$ наступления события $A$ отклонится от произведения $n \cdot p$ не более, чем на величину $\varepsilon$ примерно равна

$$P_n(|m - np| \leq \varepsilon) \approx \Phi(\frac{\varepsilon}{\sqrt{npq}})$$

б) вероятность того, что частота $m \cdot n$ события $A$ заключена в пределах от $\alpha$ до $\beta$ включительно, примерно равна


$$P_n(\alpha \leq \frac{m}{n} \leq \beta) \approx \frac{1}{2}(\Phi(z_2) - \Phi(z_1))$$

в) вероятность того, что частота $m \cdot n$ события $A$ отличается от его вероятности $p$ не более, чем на величину $\Delta > 0$ (по абсолютной величине), примерно равна

$$P_n\left(\left|\frac{m}{n} - p\right| \leq \Delta\right) \approx \Phi\left(\frac{\Delta\sqrt{n}}{\sqrt{pq}}\right)$$



### Пример

Вероятность того, что деталь нестандарта равна $0.1$. Сколько деталей надо отобрать, чтобы с вероятностью $0.9545$ можно было утверждать, что относительная частота появления нестандартных деталей отклоняется от постоянной вероятности по абсолютной величине не более чем на $0.03$

$$P_n(|\frac{m}{n} - p| \leq \Delta) \approx \Phi(\frac{\Delta\sqrt{n}}{\sqrt{pq}})$$


$$\Phi\left(\frac{0.03\sqrt{n}}{\sqrt{0.1 \cdot 0.9}}\right) = 0.9545$$

$$\Phi(0.1\sqrt{n}) = 0.9545$$

Далее по таблице Лапласа по значению функции находим значение аргумента

$$0.1\sqrt{n} = 2 \Rightarrow n = 400$$
