# 26-10-2026 Основные законы распределения ДСВ

## Закон Бернулли

> Распределение Бернулли - это дискретное распределение вероятности для СВ, которая может принимать тольку одну из двух исходов: "успех", "неудача", причем вероятность успеха $p$ заранее задано

$$P(X = 1) = p, P(X = 0) = q$$

где $0 < p < 1$, $q = 1 - p$

$$M(X) = p, D(X) = pq$$

## Биномиальное распределение

> Биномиальный закон распределения представляет собой закон распределения числа $X = m$ наступления события $A$ в $n$ независимых испытаниях, в каждом из которых оно может произойти с одной и той же вероятностью $p$.

$$P(X = m) = C^m_np^mq^{n-m}$$

$$M(X) = np, D(X) = npq$$

### Пример. 

Вероятность отказа детали за время испытания на надежность равна $0.2$. Найти закон распределения числа отказавших деталей, математическое ожидание и дисперсию, если произведено испытание $6$ деталей

## Закон распределения Пуассона

> ДСВ $X$ имеет закон распределения Пуассона, если она принимает значения $0, 1, 2, \dots, m, \dots$

$$P(X = m) = \frac{\lambda^{m} \cdot e^{-\lambda}}{m!} = P_{m}(\lambda)$$

$$M(X) = D(X) = \lambda$$

### Пример.

Пусть число покупателей, пришедших в магазин за $1$ минуту, – это случайная
величина, имеющая распределение Пуассона с параметром $\lambda = 3.5$. Событие $A$ – за $1$ минуту пришло более $1$ покупателя. Найти $P(A)$.

$$P(A) = 1 - P(X = 0) - P(X = 1) = 1 - \frac{3.5^0 \cdot e^{-3.5}}{0!} - \frac{3.5^1 \cdot e^{-3.5}}{1!} = 1 - e^{-3.5}(1+3.5)$$

### Геометрическое распределение, вычисление характеристик геометрической ДСВ

> ДСВ $X$ имеет геометрическое распределение, если она принимает значения $1, 2, \dots, m, \dots$ (бесконечное, но счетное множество значений) с вероятностями

$$P(X = m) = pq^{m - 1}$$

где $0 < p < 1$, $q = 1 - p$, $m = 1, 2, \dots$

Случайная величина, имеющая геометрическое распределение, представляет собой число $m$ испытаний, проведенных по схеме Бернулли, с вероятностью $p$ наступления события в каждом испытании до первого положительного исхода.

$$M(X) = \frac{1}{p} D(X) = \frac{q}{p^2}$$

### Пример.

Вероятность поражения цели равна $0.05$. Производится стрельба по цели до первого попадания. Необходимо: а) составить закон распределения числа сделанных выстрелов; б) найти математическое ожидание и дисперсию этой случайной в) определить вероятность того, что для поражения цели потребуется не менее $5$ выстрелов.

> ДСВ $X$ имеет **гипергеометрическое распределение** с параметрами $n$, $M$, $N$ если она принимает значения $0, 1, 2, \dots, m, \dots, min(n, M)$ с вероятностями 

$$P(X = m) = \frac{C^mC^{n-m}_{N-M}}{C^n_N}$$

где $m = 0, 1, 2, \dots, min(n, M), M \leq N, n \leq N, n, M, N$ − натуральные числа

$$M(X) = n \frac{M}{N}$$ $$D(X) = n \frac{M}{N - 1}\left(1 - \frac{M}{N}\right)\left(1 - \frac{n}{N}\right)$$

### Пример.

В магазине имеются $20$ телевизоров, из них $7$ имеют дефекты. Необходимо: а)
составить закон распределения числа телевизоров с дефектами среди выбранных наудачу $5$; б) найти математическое ожидание и дисперсию этой случайной величины; в) определить вероятность того, что среди выбранных нет телевизоров с дефектами.
