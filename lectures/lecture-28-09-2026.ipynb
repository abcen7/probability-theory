# 28.09.2026 Повторные независимые испытания. Формула Бернулли

$P(Б) = 0.01$ – болен, $P(З) = 0.99$ – здоров; $P_Б(П)=0.95$ – положительный тест того, что болен, $P_З(П) = 0.05$ – положительный тест того, что здоров

Надо найти $P_П(Б) = ?$ – болен, при положительном тесте

По формуле Байеса: $P_П(Б) = \frac{P(Б) \cdot P_Б(П)}{P(П)} = \frac{P(Б) \cdot P_Б(П)}{P(Б) \cdot P_Б(П) + P(З) \cdot P_З(П)} = \frac{0.01 \cdot 0.05}{0.01 \cdot 0.95 + 0.99 \cdot 0.05} = 0.16$

> Если производится несколько испытаний, причем вероятность события $A$ в каждом испытании **не зависит** от исходов других, то такие испытания называются **независимые** от события $A$.

> Сложные события – совмещения нескольких простых

> Теорема Бернулли. Если вероятность $p$ наступления события $A$ в каждом испытании постоянна, то вероятность $P_n(m)$ того, что событие $A$ наступит $m$ раз в $n$ испытаниях равна
$$P_n(m) = C_n^m \cdot p^m \cdot q^{n - m}$$

### Пример. Вероятность изготовления на станке стандартной детали равна $0.8$. Найти вероятности возможного появления бракованных деталей среди $5$ бракованных; 

$n = 5$

$p = 0.8$ (стандартная)

$\bar{p} = 0.2$ (бракованная)

$m = {0, 1, \dots, 5}$

- $P_5(0) = C_5^0 \cdot 0.2^0 \cdot 0.8^5 = 0.32768$
- $P_5(1) = C_5^1 \cdot 0.2^1 \cdot 0.8^4 = 0.4096$
- $P_5(2) = C_5^2 \cdot 0.2^2 \cdot 0.8^3 = 0.2048$
- $P_5(3) = C_5^3 \cdot 0.2^3 \cdot 0.8^2 = 0.0512$
- $P_5(4) = C_5^4 \cdot 0.2^4 \cdot 0.8^1 = 0.0064$
- $P_5(5) = C_5^5 \cdot 0.2^5 \cdot 0.8^0 = 0.00032$

$\sum\limits_{n = 1}^{5}P_i = 1$

> Наивероятнейшим числом наступления события $A$ называется число, вероятность которого является наибольшей из вероятностей наступления события $A$ любое другое число раз. Оно определяется по формуле $$np - q \leq m_0 \leq np + p$$

$$5 \cdot 0.2 - 0.8 \leq m_0 \leq 5 \cdot 0.2 + 0.2$$
$$0.2 \leq m_0 \leq 1.2$$

## Ассимптотические формулы оценки вероятности (приближенное вычисление схемы Бернулли для $N \to \infty$ 

### Формула Пуассона. (Закон редких явлений)

Критерий применимости: $\lambda = np \leq 10$

Если вероятность $p$ наступления события $A$ в каждом испытании стремится к нулю $(p \to 0)$ при неограниченном увеличении числа испытаний $(n \to \infty)$. Причем произведение стремится к постоянному числу $\lambda$: $np \to \lambda$, то вероятность $P_n(m)$ того, что событие $A$ появится $m$ раз в $n$ независимых испытаниях удовлетворяет предельному равенству.

$$\lim\limits_{n \to \infty}{P_n(m) = P_m(\lambda) = \frac{\lambda^m \cdot e^{-\lambda}}{m!}}$$

Если вероятность $p$ – постоянна и мала, число испытаний $n$ велико и число $\lambda = np$ – незначительна ($\lambda \leq 10$), то можно использовать приближенную формулу Пуассона: 

$$P_n(m) \approx \frac{\lambda^m \cdot e^{-\lambda}}{m!} = P_m(\lambda)$$

### Пример. На факультете насчитывается $1825$ студентов. Какова вероятность того, что первого сентября является днем рождения одновременно у $4$ студентов?

$n = 1825$

$m = 4$

$p = \frac{1}{365}$

$\lambda = np = 1825 \cdot \frac{1}{365} = 5 \leq 10 \Rightarrow $ можно использовать формулу Пуассона.

$P_{1825}(4) \approx P_4(5) = 0.1755$

## Локальная формула Муавра – Лапласа

Если вероятность $p$ наступления события $A$ в каждом испытании постоянна и отлична от 0 и 1, то вероятность $P_n(m)$ того, что событие $A$ произойдет $m$ раз в $n$ независимых испытаниях при достаточно большом числе $n$, приблизительно равна

$$P_n(m) \approx \frac{f(x)}{\sqrt{npq}}$$

где $f(x) = \frac{1}{\sqrt{2\pi}} \cdot e^{-\frac{x^2}{2}}$ - функция Гаусса. $x = \frac{m - np}{\sqrt{npq}}$. Незначительная погрешность (удовлетворительная), если $npq \geq 20$. Значение функции Гаусса можно посмотреть в таблице

Свойства функции Гаусса. $P_n(m) \approx \frac{f(x)}{\sqrt{npq}}$

- Четность. $f(x) = -f(x)$
- При $x \to \infty: f(x) \to 0$,
- При $x > 0, f(x) \approx 0$

### Пример. Найти вероятность события $A$, наступит ровно $80$ раз в $400$ испытаниях, если вероятность появления в каждом испытании равна $0.2$

- $n = 400$
- $m = 80$
- $p = 0.2$

$400 \cdot 0.2 \approx 80 \leq 10 \rightarrow$ Пуассон не подходит

Оценим $npq = 400 \cdot 0.2 \cdot 0.8 = 64 \rightarrow$ можно использовать лк М-У

$x = \frac{m - np}{\sqrt{npq}} = \frac{80 - 400 \cdot 0.2}{\sqrt{64}} = 0$

$P_{400}(80) \approx \frac{f(0)}{8} \cdot \frac{0.3989}{8} = 0.04986$

## Интегральная теорема Муавра-Лапласа

> Теорема. Есть вероятность $p$ наступления события $A$ в каждом испытании постоянна и отлична от $0$ и $1$, то вероятность того, что число $m$ наступления события $A$ в $n$ независимых испытаниях заключено в пределах от $a$ до $b$ (включительно), при достаточно большом числе $n$ = $$P_n(a \leq m \leq b) \approx \frac{1}{2}(Ф(x_2) - Ф(x_1))$$

где $$Ф(x) = \frac{2}{\sqrt{2\pi}}\int\limits^{x}_{0}e^{-\frac{t^2}{2}}dt$$ - функция Лапласа: $$x_1 = \frac{a-np}{\sqrt{npq}}, x_2 = \frac{b - np}{\sqrt{npq}}$$

Значение функции Лапласа находится по таблице. 

Свойства функции $Ф(x)$

- Функция $Ф(x)$ нечетная: $Ф(x)$ = $-Ф(x)$
- Функция $Ф(x)$ монотонно возрастающая, причем $x \to +\infty, Ф(x) \to 1$, можно считать, что уже при $x > 4: Ф(x) \approx 1$

### Пример. Вероятность того, что деталь не прошла проверку ОТК равна $0.2$. Найти вероятность того, что среди $400$ случайно отобранных деталей окажутся непроверенными от $70$ до $100$ деталей.

$n = 400$

не проверено: $a = 70, b = 100$

(не проверено): $p = 0.2$

Оценим $npq = 400 \cdot 0.2 \cdot 0.8 = 64 \geq 20 \Rightarrow$ можно применить интегральную Т М-Л

$x_1 = \frac{a - np}{\sqrt{npq}} = \frac{70 - 80}{8} = \frac{-10}{8} = -1.25$

$x_2 = \frac{b - np}{\sqrt{npq}} = \frac{100 - 80}{8} \frac{20}{8} = 2.5$

$P(70 \leq m \leq 100) \approx \frac{1}{2}(Ф(2.5) - Ф(-1.25)) \approx \frac{1}{2}(Ф(2.5) + Ф(1.25)) \approx \frac{1}{2}(0.9876 + 0.7887) = 0.88814$
