### №1.10

$P(A) = \frac{m}{n} = \frac{1}{6*5*4} = \frac{1}{120} = 0.0083$

### №1.16

а) Пусть $A$ - событие, никакие два лица одного пола не сядут рядом, где $m$ – количество благоприятных случаев:

Мужчины могут занять четные места: 5! 
Женщины могут занять нечетные места: 5!
И наоборот, следовательно: * 2

А, $n$ – количество всех случаев: $P_{10} = 10!$

$P(A) = \frac{5! * 5! * 2}{P_{10}} = \frac{2 * (5!)^2}{3628800} = 0.0079$ 

```python
from scipy.special import factorial as f
print(round(((f(5) ** 2) * 2) / f(10), 4))
```

```python
from scipy.special import perm
print(round((perm(5, 5) * perm(5, 5) * 2) / perm(10, 10), 4))
```

### №1.17

а) Пусть событие $A$ – пассажиры попали в два купе. Общее число – $n = C_{36}^7$. Количество благоприятных случаев определяется по правилу умножения выборки $2$ купе из $9$ И учета того, что все пассажиры займут $7$ мест из $8$ в выбранных $2$ купе: $C_{9}^2 * C_{8}^7$. Итого:

$P(A) = \frac{C_{9}^2 * C_{8}^7}{C_{36}^7} = 0.00003$

```python
from scipy.special import comb

result = round((comb(9, 2) * comb(8, 7)) / (comb(36, 7)), 5)

print(f'{result:.5f}')
```

б) Пусть событие $A$ – пассажиры попали в два купе. Общее число – $n = C_{36}^7$. Количество благоприятных случаев определяется по правилу умножения выборки $7$ купе из $9$ И учета того, что все пассажиры выберут $1$ место из $4$ в выбранных $2$ купе: $C_{9}^7 * 4^7$. Итого:

$P(A) = \frac{C_{9}^7 \cdot 4^7}{C_{36}^7}$

```python
from scipy.special import comb
print(round((comb(9, 7) * (4 ** 7)) / comb(36, 7), 4))
```
