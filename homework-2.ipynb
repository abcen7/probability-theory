### №3.27

Рассмотрим СВ $X$ - размер выигрыша при пяти сделанных покупках в **тысячах** рублей

- $p = 0.1$
- $q = 0.9$
- $n = 5$

$$X \sim Bin$$

$$P(X = k) = C^k_n p^k q^{n-k}, k = \overline{0, 5}$$

$$
\begin{array}{|c|c|c|c|c|c|c|}
\hline
x_i & 0 & 1 & 2 & 3 & 4 & 5 \\
\hline
p_i & 0.59049 & 0.32805 & 0.0729 & 0.0081 & 0.00045 & 0.00001 \\
\hline
\end{array}
$$

$$M(X) = np = \boxed{0.5}$$
$$D(X) = npq = \boxed{0.45}$$

```python
import numpy as np
import pandas as pd
from scipy.stats import binom

n = 5
p = 0.1
q = 1 - p

np.set_printoptions(formatter={'float_kind': '{:0.4f}'.format})

X = np.arange(0, n + 1)
binomial_distribution = binom(n, p)
probabilities = binomial_distribution.pmf(X)

df = pd.DataFrame(
    data=probabilities.reshape(1, -1),
    columns=X,
    index=['p']
)

df
```

```python
print(f'M(X) = {binomial_distribution.mean():.4f}')
print(f'D(X) = {binomial_distribution.var():.4f}')
```

```python
print(f'M(X) = {n * p}')
print(f'D(X) = {n * p * q}')
```

- $P(X = 1) = 0.9 \cdot 0.2 \cdot 0.3 + 0.1 \cdot 0.8 \cdot 0.3 + 0.1 \cdot 0.2 \cdot 0.7 = 0.0920$
- $P(X = 2) = 0.9 \cdot 0.8 \cdot 0.3 + 0.9 \cdot 0.2 \cdot 0.7 + 0.1 \cdot 0.8 \cdot 0.7 = 0.3980$

$$M(X) = \boxed{2.4}$$

### №3.40

Рассмотрим СВ $X$ - число библиотек, которые посетит студент, если в городе $4$ библиотеки

- $p = 0.3$
- $n = 4$

$$X \sim G(p)$$

- $P(X = 1) = 0.3$
- $P(X = 2) = 0.7 \cdot 0.3 = 0.21$
- $P(X = 3) = 0.7 \cdot 0.7 \cdot 0.3 = 0.147$
- $P(X = 4) = 0.7 \cdot 0.7 \cdot 0.7 = 0.3430$


$$
\begin{array}{|c|c|c|}
\hline
x_i & 1 & 2 & 3 & 4 \\
\hline
p_i & 0.3000 & 0.2100 & 0.1470 & 0.3430 \\
\hline
\end{array}
$$

$$M(X) = \boxed{2.5330}$$
$$D(X) = \boxed{1.5349}$$

```python
import numpy as np
import pandas as pd
from scipy.stats import geom

np.set_printoptions(formatter={'float_kind': '{:0.4f}'.format})
pd.options.display.float_format = '{:,.3f}'.format

p = 0.3
q = 1 - p
n = 4
x = np.arange(1, n + 1)

geom_distribution = geom(p)
probabilities = geom_distribution.pmf(x)

# (!) Потому что последнюю библиотеку студент посетит вне зависимости от результата
probabilities[-1] = geom.sf(3, p)

df = pd.DataFrame(
    data=probabilities.reshape(1, -1),
    columns=x,
    index=['p']
)

df
```

```python
mean = np.sum(x * probabilities)
variance = np.sum((x - mean) ** 2 * probabilities)

print(f'M(X) = {mean:.4f}')
print(f'D(X) = {variance:.4f}')
```

### №3.46

Рассмотрим СВ $X$ — число попыток открывания замка

$$p = \boxed{\frac{1}{4}}$$

- $P(X = 1) = \frac{1}{4}$
- $P(X = 2) = \frac{3}{4} \cdot \frac{1}{3} = \frac{1}{4}$
- $P(X = 3) = \frac{3}{4} \cdot \frac{2}{3} \cdot \frac{1}{2} = \frac{1}{4}$
- $P(X = 4) = \frac{3}{4} \cdot \frac{2}{3} \cdot \frac{1}{2} \cdot 1 = \frac{1}{4}$

$$P(X = k) = \frac{1}{4}, k = \overline{1,4}$$

$$
\begin{array}{|c|c|c|c|}
\hline
x_i & 1 & 2 & 3 & 4 \\
\hline
p_i & 0.25 & 0.25 & 0.25 & 0.25 \\
\hline
\end{array}
$$

$$M(X) = \boxed{2.50}$$
$$D(X) = \boxed{1.25}$$
$$\sigma(X) = \boxed{1.1180}$$

```python
import numpy as np
import pandas as pd
from scipy.stats import randint

distribution = randint(1, 5)
x = np.arange(1, 5)

df = pd.DataFrame({
    "X": x,
    "P(X)": distribution.pmf(x)
})

df
```

```python
print(f'M(X) = {distribution.mean():.4f}')
print(f'D(X) = {distribution.var():.4f}')
print(f'sigma(X) = {distribution.std():.4f}')
```
