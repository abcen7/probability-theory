Рассмотрим СВ $X$ - число выигрышей по $19$ облигациям

$P(X = m) = C^m_np^mq^{n-m}$

```python
import numpy as np
from scipy.stats import binom

p = 0.1
q = 1 - p
n = 19

dist = np.arange(0, n + 1)
x = np.array(list(dist))

probabilities = []

for m in dist:
    probabilities.append(binom.pmf(m, n, p))

probabilities = np.array(probabilities)
```

```python
M = np.sum(x * probabilities)
D = np.sum(x ** 2 * probabilities) - M ** 2

print("Матожидание:", M)
print("Дисперсия:", D)
```
