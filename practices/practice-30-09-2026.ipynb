#### Разбор от ИИ.

### 1. Что происходит после первого переноса

В первой урне:

* 5 белых;
* 6 чёрных;
* всего 11 шаров.

Из неё достают **3 шара**. Количество белых среди этих трёх может быть:

$$
k=0,1,2,3.
$$

После переноса во второй урне будет всего

$$
4+8+3=15
$$

шаров.

Если перенесли $k$ белых, то во второй урне станет:

* $4+k$ белых;
* $11-k$ чёрных.

---

### 2. Вероятность каждого значения $k$

Используем гипергеометрическое распределение:

$$
P(k)=\frac{C_5^k C_6^{3-k}}{C_{11}^3}.
$$

Получаем:

$$
P(0)=\frac{C_5^0C_6^3}{C_{11}^3}
=\frac{20}{165},
$$

$$
P(1)=\frac{C_5^1C_6^2}{C_{11}^3}
=\frac{75}{165},
$$

$$
P(2)=\frac{C_5^2C_6^1}{C_{11}^3}
=\frac{60}{165},
$$

$$
P(3)=\frac{C_5^3C_6^0}{C_{11}^3}
=\frac{10}{165}.
$$

### 3. Вероятность достать 4 белых из второй урны

Если перенесли \(k\) белых, то белых во второй урне \(4+k\).

Вероятность вытащить все 4 белых:

$$
P(A|k)=\frac{C_{4+k}^4}{C_{15}^4}.
$$

Поэтому по формуле полной вероятности:

$$
P(A)=
\sum_{k=0}^{3}
P(k)\cdot P(A|k).
$$

То есть

$$
P(A)=
\frac{1}{C_{11}^3C_{15}^4}
\left(
C_6^3C_4^4+
C_5^1C_6^2C_5^4+
C_5^2C_6^1C_6^4+
C_5^3C_7^4
\right).
$$

Подставляем:

$$

\frac{

$$

$$
