## Формула Пуассона

### №2.21

$p = \frac{1}{2000}$ – вероятность, что листовка "привела" клиента
$n = 100000$

а) 

$P_{100000}(48) = 50$
$q = 1 - \frac{1}{2000} = \frac{1999}{2000}$

$npq = 49.975 \geq 20 \Rightarrow$ можно использовать теорему Муавра-Лапласа

$P_{100000}(48) \approx \frac{\Phi(x)}{\sqrt{npq}}$

$x = \frac{m - np}{\sqrt{npq}} = -0.283$

$P_{100000}(48) \approx \frac{f(-0.283)}{\sqrt{49.972}} = 0.0542$

$f(x) = \frac{1}{\sqrt{2\pi}} \cdot e^{-\frac{x^2}{2}}$ – функция Гаусса

б) 

$$P_{100000}(45 \leq m \leq 55) \approx \frac{1}{2}(\Phi(x_2) - \Phi(x_1))$$

$$\Phi(x) = \frac{2}{\sqrt{2\pi}}\int\limits^{x}_{0}e^{-\frac{t^2}{2}dt}$$

$$x_1 = \frac{a-np}{\sqrt{npq}} = \frac{45-50}{\sqrt{49.975}} = -0.7073$$

$$x_2 = \frac{b-np}{\sqrt{npq}} = \frac{45-50}{\sqrt{49.975}} = 0.7073$$


$$P_{100000}(45 \leq m \leq 55) \approx \frac{1}{2}(\Phi(0.7073) - \Phi(-0.7073)) = \frac{1}{2}(\Phi(0.7073) + \Phi(0.7073)) = \Phi(0.7073) = 0.5206$$
