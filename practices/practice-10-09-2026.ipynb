## Классическая и геометрическая вероятность

а) Рассмотрим событие $A$ – карточки вынимаются в порядке следования букв слова "событие". 
$P(A) = \frac{m}{n} = \frac{m}{P_7} = \frac{1}{7!}$

```python
from scipy.special import factorial as f
print(round(1 / f(7), 4))
```

б) Рассмотрим событие $B$ – на карточках получилось слова "статистика". $P(B) = \frac{m}{n} = \frac{1}{P_7(2, 3, 2, 2, 1)} = \frac{2! \cdot 3! \cdot 2! \cdot 2!}{10!}$

```python
print(f'{(f(2) * f(3) * f(2) * f(2)) / f(10):.5f}')
```

а) $P(A) = \frac{m}{n} = \frac{C_{15}^4}{C_{25}^4}$

```python
from scipy.special import comb
round(comb(15, 4) / comb(25, 4), 4)
```

б) $P(B) = \frac{C_{10}^4}{C_{25}^4}$

```python
round(comb(10, 4) / comb(25, 4), 4)
```

в) $P(D) = \frac{m}{n} = \frac{C_{10}^3 \cdot C_{15}^1}{C_{25}^4}$

```python
round(comb(10, 3) * comb(15, 1) / comb(25, 4), 4)
```
