### Классическое определение вероятности

Рассмотрим событие $A$ – студент сдаст зачет.

$P(A) = \frac{C_{19}^2 \cdot 5 + C_{19}^3}{C_{24}^3}$

```python
from scipy.special import comb
round((comb(19, 2) * 5 + comb(19, 3)) / comb(24, 3), 4)
```

Рассмотрим событие $B$ – студент не сдаст зачет.

$B = \bar{A}$\
$P(B) = \bar{P(A)} = 1 - P(A) = 1 - 0.9092 = 0.0988$

```python
from scipy.special import comb
1 - round((comb(19, 2) * 5 + comb(19, 3)) / comb(24, 3), 4)
```
