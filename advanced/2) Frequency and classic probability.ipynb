# Тема №2. Моделирование связи частотного и классического определения вероятности

### Задача аналогичная ноутбуку №1. **Выберем 3 случайные точки на окружности. Какова вероятность, что центр окружности будет внутри треугольника, образованного этими точками?**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import math
from matplotlib.patches import Circle, Polygon
from random import randint, uniform

circle_radius = 3
circle_x0 = 0
circle_y0 = 0

def generate_random_alpha() -> float:
    return uniform(0, 2 * math.pi)

def point_in_triangle(point, triangle_vertices):
    a, b, c = triangle_vertices
    px, py = point

    def cross(a, b, p):
        return (
            (b[0] - a[0]) * (p[1] - a[1])
            - (b[1] - a[1]) * (p[0] - a[0])
        )

    s1 = cross(a, b, point)
    s2 = cross(b, c, point)
    s3 = cross(c, a, point)

    return (
        (s1 >= 0 and s2 >= 0 and s3 >= 0)
        or
        (s1 <= 0 and s2 <= 0 and s3 <= 0)
    )

def generate_random_point(alpha: float) -> tuple[float, float]:
    x = circle_x0 + circle_radius * math.sin(alpha)
    y = circle_y0 + circle_radius * math.cos(alpha)
    return x, y

def simulate(count_tries: int) -> float:
    favorable_counter = 0

    for _ in range(count_tries):
        triangle_vertices = [
            generate_random_point(generate_random_alpha())
            for _ in range(3)
        ]

        if point_in_triangle(
            (circle_x0, circle_y0),
            triangle_vertices
        ):
            favorable_counter += 1


    probability = favorable_counter / count_tries

    return probability

values = []
for _ in range(10):
    values.append(simulate(100000))

df = pd.DataFrame(
    {
        "Метод": ["Аналитический", "Симуляция"],
        "Вероятность": [1 / 4, np.mean(values)],
    }
)

df
```
